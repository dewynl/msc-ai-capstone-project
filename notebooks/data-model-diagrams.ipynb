# Educational Syllabus Generation Data Model - Visual Diagrams

This notebook renders all the Mermaid diagrams from `docs/data-model-diagram.md` using the mermaid-py library.


````python
import os
import re

from IPython.display import display
from mermaid import Mermaid
````

````python
def extract_mermaid_diagrams(md_file_path):
    """
    Extract all mermaid code blocks from a markdown file.
    Returns a list of tuples: (section_title, mermaid_code)
    """
    with open(md_file_path, encoding="utf-8") as f:
        content = f.read()

    # Split content into sections
    sections = re.split(r"^(##\s+.*?)$", content, flags=re.MULTILINE)

    diagrams = []
    current_title = "Overview"

    for section in sections:
        # Check if this section is a title
        if re.match(r"^##\s+", section):
            current_title = section.strip().replace("##", "").strip()
        else:
            # Find mermaid code blocks in this section
            mermaid_blocks = re.findall(r"```mermaid\n(.*?)\n```", section, re.DOTALL)
            for block in mermaid_blocks:
                diagrams.append((current_title, block.strip()))

    return diagrams
````

````python
def render_mermaid_diagram(title, mermaid_code):
    """
    Render a mermaid diagram with title.
    """
    print(f"\n{'='*60}")
    print(f"📊 {title}")
    print(f"{'='*60}")

    try:
        # Create Mermaid object and render
        mermaid = Mermaid(mermaid_code)
        # Note: mermaid-py renders to SVG by default
        display(mermaid)
    except Exception as e:
        print(f"❌ Error rendering diagram: {e}")
        print("\n📝 Mermaid code:")
        print("```mermaid")
        print(mermaid_code)
        print("```")
````

````python
# Load and extract diagrams from the markdown file
md_file_path = "../docs/data-model-diagram.md"

if os.path.exists(md_file_path):
    diagrams = extract_mermaid_diagrams(md_file_path)
    print(f"📋 Found {len(diagrams)} Mermaid diagrams in the markdown file.")
    print("\nDiagram sections:")
    for i, (title, _) in enumerate(diagrams, 1):
        print(f"{i}. {title}")
else:
    print(f"❌ Markdown file not found: {md_file_path}")
````
